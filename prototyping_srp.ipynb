```python
import os

import json
import pandas as pd
import requests
from dotenv import load_dotenv

load_dotenv('local.env')
BASE_URL = "https://myaccount.srpnet.com/myaccountapi/api/"

ACCOUNT_ID = os.getenv("SRP_ACCOUNT_ID")
USERNAME = os.getenv("SRP_USERNAME")
PASSWORD = os.getenv("SRP_PASSWORD")
```

```python
auth_res = requests.post(
    f"{BASE_URL}login/authorize",
    data={
         "username": USERNAME,
         "password": PASSWORD
     }
    )
if auth_res.status_code == 200:
    print(auth_res.content)
# print(auth_res.status_code)
# print(auth_res.reason)
```

```python
token_response = requests.get(
    f"{BASE_URL}login/antiforgerytoken",
    cookies=auth_res.cookies,
    data={
         "username": USERNAME,
         "password": PASSWORD
     }
    )
print(token_response.status_code)
# print(response.reason)
```
